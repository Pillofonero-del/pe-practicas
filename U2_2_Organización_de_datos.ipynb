> Llena cada celda con su valor correspondiente, los valores actuales son solo de ejemplo

```python
import pandas as pd

tiempos = [8, 7, 9, 8, 10, 7, 8, 9, 8, 7, 10, 8, 9, 7, 8]

tiempos = pd.Series(tiempos)
frecuencia_absoluta = tiempos.value_counts().sort_index()
print(frecuencia_absoluta)



# Calcula cada frecuencia
frecuencia_relativa = frecuencia_absoluta / len(tiempos)

frecuencia_acumulada = frecuencia_absoluta.cumsum()

# Creamos la tabla con las columnas correspondientes
tabla_frecuencias = pd.DataFrame({
  "Frecuencia absoluta": frecuencia_absoluta,
  "Frecuencia relativa": frecuencia_relativa,
  "Frecuencia acumulada": frecuencia_acumulada,
})

# Imprimimos los resultados de la tabla
print(tabla_frecuencias)
print(frecuencia_relativa)
print(frecuencia_acumulada)
```
